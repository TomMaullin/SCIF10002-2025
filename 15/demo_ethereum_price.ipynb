# Demo: Displaying Cryptocurrency Data using Plotly

This notebook demonstrates the capabilities of `plotly` for visualizing and interacting with complex time-series data. To do so, we will download live price data for the cryptocurrency Ethereum, denominated in GBP, and create a [candlestick chart](https://en.wikipedia.org/wiki/Candlestick_ch_)


## Setup

To run the code we must first `import` the below packages. **Note:** The `pip install` command might take a minute or two to run!

```python
!pip install requests pandas plotly

import requests
import pandas as pd
import plotly.graph_objects as go

import plotly.io as pio
pio.renderers.default = "notebook_connected"
```

## Download Data

The below code will download some live crypto data for us to look at.

```python
# Fetch some crypto data from the cracken crypto exchange
url = "https://api.kraken.com/0/public/OHLC"
params = {"pair": "ETHGBP", "interval": 5}

# Saving the data in a useful format.
data = requests.get(url, params=params).json()
rows = list(data["result"].values())[0]
```

## Preprocessing

Using `Pandas`, we will do some "light-touch" preprocessing to make sure the data is in the correct format.

```python
# Light touch data processing in Pandas
df = pd.DataFrame(rows, columns=[
    "time","open","high","low","close","vwap","volume","count"
])

df["time"] = pd.to_datetime(df["time"], unit="s")
df[["open","high","low","close"]] = df[["open","high","low","close"]].astype(float)

```

### Plotly Candlestick Chart

We now use `plotly` to create a candlestick chart for this data. Note how similar the syntax is to that of `matplotlib`. 

```python
# Create figure
fig = go.Figure(go.Candlestick(
    x=df["time"],
    open=df["open"],
    high=df["high"],
    low=df["low"],
    close=df["close"]
))

# Update the layout
fig.update_layout(
    title="Live ETH/GBP from Kraken",
    xaxis_title="Time",
    yaxis_title="Price (GBP)"
)

# Show the output
fig.show()
```

Try moving the cursor over the figure. You will see that you can drag the time window, zoom in and out, and hover over individual data points to view detailed price information.
